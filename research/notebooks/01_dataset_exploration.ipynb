```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().resolve().parent.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from ai.shared.datasets.base_dataset import AgriculturalDataset
from ai.shared.evaluation.visualization import plot_class_distribution

print("Modules imported successfully.")
```

## 1. Load Dataset Manifest

```python
manifest_dir = project_root / "data" / "manifests"
supported_crops_path = manifest_dir / "supported_crops.json"
governorates_path = manifest_dir / "yemen_governorates.json"

if supported_crops_path.exists():
    crops_df = pd.read_json(supported_crops_path)
    print(f"Loaded {len(crops_df)} supported crops.")
    display(crops_df.head())

if governorates_path.exists():
    gov_df = pd.read_json(governorates_path)
    print(f"Loaded {len(gov_df)} governorates.")
    display(gov_df.head())
```
