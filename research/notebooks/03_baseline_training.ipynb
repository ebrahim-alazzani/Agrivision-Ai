# AgriVision AI — 03 Baseline Model Training

Trains the baseline convolutional transfer learning model (ResNet50 / MobileNetV3) on the plant pathology dataset with geographic split validation.

```python
import sys
from pathlib import Path
import yaml

project_root = Path.cwd().resolve().parent.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from ai.training.classification.config import TrainingConfig
from ai.training.pipelines.experiment import Experiment
from ai.engines.plant_vision.config import PlantVisionConfig

config_path = project_root / "research" / "configs" / "baseline_config.yaml"
with open(config_path, "r", encoding="utf-8") as f:
    cfg = yaml.safe_load(f)
print(f"Loaded baseline configuration: {cfg.get('experiment_name')}")
```

## 1. Configure and Launch Training Experiment

```python
train_cfg = TrainingConfig(
    architecture="resnet50",
    epochs=5,
    batch_size=16,
    learning_rate=1e-4,
    experiment_name="notebook_baseline_demo",
)
plant_cfg = PlantVisionConfig(architecture="resnet50")

exp = Experiment(
    experiment_name="notebook_baseline_demo",
    training_config=train_cfg,
    plant_config=plant_cfg,
)
print(f"Experiment runner prepared. Output directory: {exp.output_dir}")
```
