Detailed evaluation of trained checkpoints:

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

project_root = Path.cwd().resolve().parent.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from ai.shared.evaluation.metrics import compute_classification_metrics, compute_calibration_metrics
from ai.shared.evaluation.visualization import plot_confusion_matrix

print("Evaluation modules loaded.")
```

## 1. Compute Test Metrics and Confusion Matrix

```python
classes = ["healthy", "early_blight", "late_blight", "leaf_rust"]
y_true = np.random.choice(len(classes), 100)
y_pred = y_true.copy()
# Add 10% errors
error_idx = np.random.choice(100, 10, replace=False)
y_pred[error_idx] = (y_pred[error_idx] + 1) % len(classes)

metrics = compute_classification_metrics(y_true, y_pred, class_names=classes)
print(f"Accuracy: {metrics.accuracy:.4f}")
print(f"Macro F1: {metrics.macro_f1:.4f}")

fig = plot_confusion_matrix(metrics.confusion_matrix, class_names=classes, title="Test Set Confusion Matrix")
plt.show()
```
